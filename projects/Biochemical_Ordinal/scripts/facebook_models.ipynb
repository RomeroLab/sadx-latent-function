```python
model, alphabet = esm.pretrained.esm1v_t33_650M_UR90S_1()
```

```python
batch_converter = alphabet.get_batch_converter()
```

```python
import pandas as pd
model_seqs = pd.read_csv("../output/2D/azid_model_sequences.csv")
```

```python
# select first few proteins and drop the stop character
data = [("protein" + str(i), seq) for i, seq in enumerate(model_seqs.sequence_aa.str.slice(0, -1))]
data[:2]
```

```python
batch_labels, batch_strs, batch_tokens = batch_converter(data)
```

```python
# Extract per-residue representations (on CPU)
with torch.no_grad():
    results = model(batch_tokens, repr_layers=[33], return_contacts=False)
token_representations = results["representations"][33]
```

```python
results["representations"][33].shape
```

```python
# Generate per-sequence representations via averaging
# NOTE: token 0 is always a beginning-of-sequence token, so the first residue is token 1.
sequence_representations = []
for i, (_, seq) in enumerate(data):
    sequence_representations.append(token_representations[i, 1 : len(seq) + 1].mean(0))
```

```python
torch.save(sequence_representations, "../cache/working/azid_sequence_representations.pt")
```
