```python
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../output/ordinal_Oct22_models/saved_models/6eeae50e.Hbin_probs.tsv", sep="\t")
```

```python
df["aa"] = df.feature.str.get(-1)
df["site"] = df.feature.str.slice(1, -1).astype(int)
```

```python
df
```

```python
hbin_df = df.pivot(columns="site", index="aa", values="Hbin_probs")
```

```python
fig, ax = plt.subplots(figsize=(40,5))        
sns.heatmap(hbin_df, square=True, ax=ax, cmap="coolwarm")

plt.savefig("../output/ordinal_Oct22_models/saved_models/6eeae50e.Hbin_probs.svg")
```
