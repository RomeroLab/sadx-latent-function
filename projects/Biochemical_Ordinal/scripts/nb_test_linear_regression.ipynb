```python
relaxed_column_names = [x for x in dm.column_names if x.startswith("r_")]
docking_column_names = [x for x in dm.column_names if x not in relaxed_column_names]
```

```python
def annotate_facet_with_correlations(predictions, values, ax=None, *args, **kwargs):
    pearson_r = sp.stats.pearsonr(predictions, values)
    spearman_r = sp.stats.spearmanr(predictions, values)
    if ax is None:
        ax = plt.gca()
    ax.text(0.05, 0.8, f"{pearson_r[0]:.2f} : pearson_r\n"
                   f"{spearman_r[0]:.2f}: spearman_r", 
        horizontalalignment='left',
        bbox=dict(boxstyle="round", facecolor='red', alpha=0.5),
        transform=ax.transAxes)
    
def plot_columns(column_names, dm=dm, predictions=predictions, values=values, col_wrap=5):
    nrows=len(column_names) // col_wrap
    fig, axs = plt.subplots(nrows=nrows, ncols=col_wrap, figsize=(3*col_wrap, (3.2)*nrows))
    for i, ax in enumerate(axs.flatten()):
        if i < len(column_names):
            col_name = column_names[i]
            col_idx = dm.column_names.index(col_name)
            sns.regplot(x=predictions[:, i], y=values[:, i], ax=ax)
            ax.set_title(col_name)
            annotate_facet_with_correlations(predictions=predictions[:, i],  
                                             values=values[:, i],
                                             ax=ax)
        
    plt.subplots_adjust()
```

```python
plot_columns(relaxed_column_names)
```

```python
plot_columns(docking_column_names)
```
